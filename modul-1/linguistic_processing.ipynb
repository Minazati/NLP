```python
import spacy

nlp = spacy.load("xx_ent_wiki_sm")
```

```python
#Memproses Kalimat
doc = nlp("Mahasiswa belajar Natural Language Processing.")

for token in doc:
    print(f"{token.text:<15} {token.pos_:<10} {token.dep_}")
```

```python
#Analisis POS Token
doc = nlp("Mahasiswa belajar Natural Language Processing.")

print(f"{'Token':<15} {'POS':<10} {'Lemma'}")
print("-" * 40)
for token in doc:
    print(f"{token.text:<15} {token.pos_:<10} {token.lemma_}")
```

```python
#Membuat Tabel Analisis POS
import pandas as pd

def analyze_pos(text, nlp_model):
    doc = nlp_model(text)
    results = []
    for token in doc:
        results.append({
            "Token": token.text,
            "POS": token.pos_,
            "POS Tag": token.tag_,
            "Lemma": token.lemma_,
            "Dependency": token.dep_,
            "Head": token.head.text
        })
    return pd.DataFrame(results)

df_pos = analyze_pos("Mahasiswa belajar NLP.", nlp)
print(df_pos.to_string(index=False))
```

```python
#Analisis Beberapa Kalimat
sentences = [
    "Mahasiswa belajar NLP.",
    "Dosen mengajar Natural Language Processing.",
    "Mahasiswa membaca buku.",
    "Sistem memproses data teks.",
    "Model bahasa memprediksi kata berikutnya."
]

for sentence in sentences:
    print(f"\nKalimat: {sentence}")
    df_result = analyze_pos(sentence, nlp)
    print(df_result[["Token", "POS", "Dependency", "Head"]].to_string(index=False))
    print()
```

```python
#Dependency Parsing
doc = nlp("Mahasiswa belajar NLP.")

print(f"{'Token':<15} {'Dep':<12} {'Head'}")
print("-" * 40)
for token in doc:
    print(f"{token.text:<15} {token.dep_:<12} {token.head.text}")
```

```python

# Visualisasi Dependency
from spacy import displacy

doc = nlp("Mahasiswa belajar Natural Language Processing.")

# Menampilkan visualisasi di notebook
displacy.render(
    doc,
    style="dep",
    jupyter=True,
    options={"distance": 90}
)

# Simpan sebagai HTML
html = displacy.render(
    doc,
    style="dep",
    jupyter=False,
    page=True
)

with open("dependency_tree.html", "w", encoding="utf-8") as f:
    f.write(html)

print("Dependency tree berhasil disimpan sebagai HTML.")
```

```python

# Analisis Dataset
import pandas as pd

df = pd.read_csv("../dataset/dataset_preprocessed.csv")

# Sesuaikan dengan nama kolom yang tersedia
text_column = "content_article"  # Ganti dengan nama kolom teks yang sesuai

# Ambil sample untuk analisis
sample = df.sample(min(10, len(df)), random_state=42)

for _, row in sample.iterrows():
    text = str(row[text_column])[:200]
    doc = nlp(text)

    pos_counts = {}

    for token in doc:
        pos = token.pos_
        pos_counts[pos] = pos_counts.get(pos, 0) + 1

    print(f"Teks: {text[:80]}...")
    print(f"POS Distribution: {pos_counts}")
    print()
```
