```python
#Membuat Tabel Analisis POS
import pandas as pd

def analyze_pos(text, nlp_model):
    doc = nlp_model(text)
    results = []
    for token in doc:
        results.append({
            "Token": token.text,
            "POS": token.pos_,
            "POS Tag": token.tag_,
            "Lemma": token.lemma_,
            "Dependency": token.dep_,
            "Head": token.head.text
        })
    return pd.DataFrame(results)

df_pos = analyze_pos("Mahasiswa belajar NLP.", nlp)
print(df_pos.to_string(index=False))
```

```python
#Analisis Beberapa Kalimat
sentences = [
    "Mahasiswa belajar NLP.",
    "Dosen mengajar Natural Language Processing.",
    "Mahasiswa membaca buku.",
    "Sistem memproses data teks.",
    "Model bahasa memprediksi kata berikutnya."
]

for sentence in sentences:
    print(f"\nKalimat: {sentence}")
    df_result = analyze_pos(sentence, nlp)
    print(df_result[["Token", "POS", "Dependency", "Head"]].to_string(index=False))
    print()
```

```python
import spacy
from spacy import displacy

# Load model yang memiliki dependency parser
nlp = spacy.load("en_core_web_sm")

# Memproses teks
doc = nlp("Mahasiswa belajar Natural Language Processing.")

# Menampilkan visualisasi di notebook
displacy.render(
    doc,
    style="dep",
    jupyter=True,
    options={"distance": 90}
)

# Simpan sebagai HTML
html = displacy.render(
    doc,
    style="dep",
    jupyter=False,
    page=True
)

with open("dependency_tree.html", "w", encoding="utf-8") as f:
    f.write(html)

print("Dependency tree berhasil disimpan sebagai HTML.")
```

```python
# Analisis Dataset
import pandas as pd

df = pd.read_csv("dataset_processed.csv")

# Sesuaikan dengan nama kolom yang tersedia
text_column = "content_article"  # Ganti dengan nama kolom teks yang sesuai

# Ambil sample untuk analisis
sample = df.sample(min(10, len(df)), random_state=42)

for _, row in sample.iterrows():
    text = str(row[text_column])[:200]
    doc = nlp(text)

    pos_counts = {}

    for token in doc:
        pos = token.pos_
        pos_counts[pos] = pos_counts.get(pos, 0) + 1

    print(f"Teks: {text[:80]}...")
    print(f"POS Distribution: {pos_counts}")
    print()
```

```python
import pandas as pd
from spacy import displacy

df = pd.read_csv("dataset_processed.csv")
df = df[df["content_article"] != "Format Non-Text/Gagal Ekstrak"]

boilerplate = ["Reporter", "Editor", "KONTAN", "CNBC", "Bisnis.com", "Baca juga", "|"]

kalimat = []
for teks in df["content_article"].sample(frac=1, random_state=42):   # acak seluruh artikel
    doc_teks = nlp(str(teks)[:3000])
    for sent in doc_teks.sents:
        s = sent.text.strip()
        if 6 <= len(sent) <= 18 and not any(b in s for b in boilerplate) and s not in kalimat:
            kalimat.append(s)
            break                      # 1 kalimat per artikel
    if len(kalimat) == 10:
        break

print("Jumlah kalimat terpilih:", len(kalimat))

baris = []
for i, k in enumerate(kalimat, 1):
    doc = nlp(k)
    baris.append({
        "Kalimat": k,
        "Token": [t.text for t in doc],
        "POS": [t.pos_ for t in doc],
        "Dependency": [f"{t.text}:{t.dep_}->{t.head.text}" for t in doc],
        "Kesalahan?": ""   # isi manual
    })
    html = displacy.render(doc, style="dep", jupyter=False, page=True)
    with open(f"dep_{i}.html", "w", encoding="utf-8") as f:
        f.write(html)

hasil = pd.DataFrame(baris)
hasil.to_csv("tabel_analisis_linguistik.csv", index=False)
hasil
```
