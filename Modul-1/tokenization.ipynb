```python
import pandas as pd
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize

nltk.download("punkt")
nltk.download("punkt_tab")
```

```python
# Load Dataset
data = pd.read_csv("dataset.csv")
text_column = "content_article"
print("Jumlah data:", len(data))

print(data.head())
```

```python
# Sentence Segmentation
text = "Saya belajar NLP. NLP sangat menarik. Saya ingin menjadi NLP Engineer."

sentences = sent_tokenize(text, language='english')  # Ganti 'english' dengan bahasa yang sesuai jika diperlukan

for sentence in sentences:
    print(sentence)
```

```python
#Word Tokenization
from nltk.tokenize import word_tokenize

text = "Saya belajar Natural Language Processing."

tokens = word_tokenize(text)

print(tokens)
```

```python
#Character Tokenization
text = "NLP"

characters = list(text)

print(characters)
```

```python
# Tokenization Dataset
data["tokens"] = data[text_column].astype(str).apply(word_tokenize)

data[[text_column, "tokens"]].head()

#Menghitung jumlah token per dokumen
data["jumlah_token"] = data["tokens"].apply(len)
data["jumlah_token"].describe()
```

```python
#Membandingkan Kata dan Token
data["jumlah_kata"] = data[text_column].astype(str).apply(lambda x: len(x.split()))
data["jumlah_token"] = data["tokens"].apply(len)
data[["jumlah_kata", "jumlah_token"]].describe()
```

```python
#Eksperimen Tokenisasi
texts = [
    "Saya suka produk ini.",
    "Bagus banget!!!",
    "Tidak bagus...",
    "Harga Rp50.000.",
    "email@example.com",
    "https://example.com",
    "yg bagus bgt sih",
    "aku udah beli nih kak"
]

for text in texts:
    print("TEXT :", text)
    print("TOKEN:", word_tokenize(text))
    print()
```

```python
# Perbandingan Tokenizer
import pandas as pd
import nltk
import spacy
from nltk.tokenize import word_tokenize

# Load model multilingual
nlp = spacy.load("xx_ent_wiki_sm")

# Daftar teks untuk dibandingkan
teks_list = [
    "Saya belajar NLP.",
    "Bagus banget!!!",
    "Harga Rp50.000"
]

# Membandingkan hasil tokenizer
hasil = []

for teks in teks_list:
    nltk_tokens = word_tokenize(teks)
    split_tokens = teks.split()
    doc = nlp(teks)
    spacy_tokens = [token.text for token in doc]

    hasil.append({
        "Teks": teks,
        "NLTK word_tokenize": str(nltk_tokens),
        "split()": str(split_tokens),
        "spaCy": str(spacy_tokens)
    })

# Membuat tabel perbandingan
tabel_perbandingan = pd.DataFrame(hasil)

tabel_perbandingan
```
